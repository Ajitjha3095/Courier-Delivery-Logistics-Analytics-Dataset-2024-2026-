```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("courier_raw_data_2024_2026.csv")
```

```python
df.shape
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"])
df["Delivery_Date"] = pd.to_datetime(df["Delivery_Date"])
```

```python
(df["Delivery_Date"] < df["Order_Date"]).sum()
```

```python
df = df.drop_duplicates(subset="Shipment_ID")
```

```python
text_columns = [
    "Customer_Type",
    "Origin_Zone",
    "Destination_Zone",
    "Package_Type",
    "Shipping_Mode",
    "Delivery_Status"
]

for col in text_columns:
    df[col] = df[col].str.strip()
```

```python
(df["Shipping_Cost"] <= 0).sum()
```

```python
df["Delivery_Days"] = (
    df["Delivery_Date"] - df["Order_Date"]
).dt.days
```

```python
df["Delivery_Performance"] = np.where(
    (df["Delivery_Status"] == "Delivered") &
    (df["Delivery_Days"] <= 3),
    "On Time",
    "Not On Time"
)
```

```python
df = df[
    [
        "Shipment_ID",
        "Order_Date",
        "Delivery_Date",       
        "Customer_Type",
        "Origin_Zone",
        "Destination_Zone",
        "Package_Type",
        "Shipping_Mode",
        "Shipping_Cost",
        "Delivery_Days",
        "Delivery_Status",
        "Delivery_Performance"
    ]
]
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("Missing Values:", df.isnull().sum().sum())
print("Duplicate Shipments:", df["Shipment_ID"].duplicated().sum())
```

```python
df.to_csv(
    "courier_cleaned_data_2024_2026.csv",
    index=False
)
```
