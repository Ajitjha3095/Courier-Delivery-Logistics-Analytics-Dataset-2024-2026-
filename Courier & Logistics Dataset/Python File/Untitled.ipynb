```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(42)

n = 2000

# 1. Shipment ID
shipment_id = [f"SHP{str(i).zfill(5)}" for i in range(1, n + 1)]

# 2. Order Date
order_date = pd.to_datetime(
    np.random.choice(
        pd.date_range("2024-01-01", "2026-12-31"),
        size=n
    )
)

# 3. Delivery Date
delivery_days = np.random.choice(
    [1, 2, 3, 4, 5, 6, 7],
    size=n,
    p=[0.15, 0.25, 0.25, 0.15, 0.10, 0.06, 0.04]
)

delivery_date = order_date + pd.to_timedelta(delivery_days, unit="D")

# 4. Customer Type
customer_type = np.random.choice(
    ["Individual", "Business"],
    size=n,
    p=[0.65, 0.35]
)

# 5. Origin Zone
origin_zone = np.random.choice(
    ["North", "South", "East", "West", "Central"],
    size=n
)

# 6. Destination Zone
destination_zone = np.random.choice(
    ["North", "South", "East", "West", "Central"],
    size=n
)

# 7. Package Type
package_type = np.random.choice(
    ["Parcel", "Document", "Fragile", "Electronics"],
    size=n,
    p=[0.45, 0.20, 0.20, 0.15]
)

# 8. Shipping Mode
shipping_mode = np.random.choice(
    ["Standard", "Express", "Same Day"],
    size=n,
    p=[0.55, 0.35, 0.10]
)

# 9. Shipping Cost
base_cost = np.random.randint(80, 501, size=n)

shipping_cost = np.where(
    shipping_mode == "Express",
    base_cost * 1.4,
    np.where(
        shipping_mode == "Same Day",
        base_cost * 1.8,
        base_cost
    )
)

shipping_cost = np.round(shipping_cost).astype(int)

# 10. Delivery Status
delivery_status = np.random.choice(
    ["Delivered", "Delayed", "Cancelled", "Failed"],
    size=n,
    p=[0.78, 0.12, 0.06, 0.04]
)

# Create DataFrame
courier_df = pd.DataFrame({
    "Shipment_ID": shipment_id,
    "Order_Date": order_date,
    "Delivery_Date": delivery_date,
    "Customer_Type": customer_type,
    "Origin_Zone": origin_zone,
    "Destination_Zone": destination_zone,
    "Package_Type": package_type,
    "Shipping_Mode": shipping_mode,
    "Shipping_Cost": shipping_cost,
    "Delivery_Status": delivery_status
})

# Sort by Order Date
courier_df = courier_df.sort_values("Order_Date").reset_index(drop=True)

# Display first 10 rows
courier_df.head(10)
```
