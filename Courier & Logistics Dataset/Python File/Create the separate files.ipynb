```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("courier_cleaned_data_2024_2026.csv")
```

```python
np.random.seed(42)

df["Customer_ID"] = [
    f"CUST{str(i).zfill(4)}"
    for i in np.random.randint(1, 501, size=len(df))
]

df.head()
```

```python
customers = df[
    ["Customer_ID", "Customer_Type"]
].drop_duplicates()

customers.to_csv("customers.csv", index=False)

customers.head()
```

```python
shipments = df[
    [
        "Shipment_ID",
        "Customer_ID",
        "Order_Date",
        "Delivery_Date",
        "Origin_Zone",
        "Destination_Zone",
        "Package_Type",
        "Shipping_Mode",
        "Shipping_Cost",
        "Delivery_Status",
        "Delivery_Days",
        "Delivery_Performance"
    ]
]

shipments.to_csv("shipments.csv", index=False)

shipments.head()
```

```python
zones = pd.DataFrame({
    "Zone_ID": ["Z01", "Z02", "Z03", "Z04", "Z05"],
    "Zone_Name": [
        "North",
        "South",
        "East",
        "West",
        "Central"
    ]
})

zones.to_csv("zones.csv", index=False)

zones
```
